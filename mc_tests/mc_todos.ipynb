
# Some explorations of Monte Carlo

As outlined in the lecture, MC techniques are a bit a fringe tool in biomolecular simulation. There is little that can be done with MC, which cannot be done with MD as well, and equally well. This unit, therefore, is mainly intended to illustrate some of the move types used in biomolecular MC. Hybrid methods, combinations of MC and MD are not covered.

The following input files are in the folder:

* <tt>mc_setup.inp</tt>
* <tt>mc_test_move.inp</tt>
* <tt>mc_opt_move.inp</tt>
* <tt>mc_prod_test.inp</tt>

Here you can find the full [documentation of the MC module of CHARMM](https://www.charmm.org/charmm/documentation/by-version/c45b1/mc.html); it's unfortunately reference material only.

### 1. Getting started

Here, <tt>mc_setup.inp</tt> just prepares PSF and starting coordinates for the following jobs (no MC in there, you just have to run it first.)

    charmm -i mc_setup.inp > mc_setup.out

The model system is a tetrapetide, RGDS (Arg-Gly-Asp-Ser) in zwitterionic form. We do our tests in gas phase, where there are strong electrostatic interactions between the base (Arg) and the acid (Asp), as well as the deprotonated C-terminus. The starting coordinates, which are contructed when running <tt>mc_setup.inp</tt> are a linear chain. (This unit uses the CHARMM22 force field; CHARMM36/36m is the modern default, but for illustrating MC move types the choice makes no real difference.)


### 2. Testing move sets

The main script to play with is <tt>mc_test_move.inp</tt>. Call charmm as

      charmm movetype:x -i mc_test_move.inp > x.out
      
where x = 1, 2, ...7. Each X value activates one of the following move types.

1) single atom displacement
2) (naive) rotation, dihedrals in side chains only
3) dihedral move, $\Psi$ angle
4) dihedral move, $\Phi$ angle
5) dihedral move, $\omega$ angle
6) concerted rotation (7 backbone angles with rigid $\omega$)
7) like 6), but flexible $\omega$

Based on the value X in <tt>movetype:X</tt> passed via the command line, the script first sets up this movetype, then it carries out 1000 steps of MC, writing coordinates to a trajectory at every step. You can then download these to your computer/laptop and visualize them using VMD. NOTE: For X=1,2,3,4,5 the simulation is started from the extended chain conformation generated in 'Getting started'. The concerted rotations "do nothing" for this conformation, so for X=6,7 coordinates after some MC steps are read, and the 1000 MC steps are started from there (so what you'll see for X=6,7 looks a bit different than for X=1--5!)

**Some hints:** As you know, to visualize a trajectory (DCD file), you also need to load a PSF file. The PSF file to use is in

    ./data/rgds.psf # relative to the MC folder in your JH work directory
    
Similarly, the trajectories generated by <tt>mc_test_move.inp</tt> are written to

    ./data/mc_test_move.X.dcd   # relative to the MC folder in your JH work directory
    
Finally, once you have downloaded PSF and DCD to a working directory on your machine, *at least on Linux* you can load them into VMD simultaneously from the command line

    vmd rgds.psf mc_test_move.X.dcd  # in the directory on your computer where you downloaded PSF and DCDs
    
Whether this works when starting VMD from the command line under Windows / MacOSX as well, I do not know.



### 3. Mini production example

The [MC module in CHARMM](https://www.charmm.org/charmm/documentation/by-version/c45b1/mc.html) automatically optimizes move types / sets. To illustrate the idea, run the following sequence

     charmm -i mc_opt_move.inp > mc_opt_move.out
     charmm -i mc_prod_test.inp > mc_prod_test.noopt.out
     charmm opt:1 -i mc_prod_test.inp > mc_prod_test.opt.out
     
The first run carries out 200k MC steps using all 7 movetypes you explored earlier and optimizes the (maximum 'step-width' etc); it saves the optimized moves to a file. The two calculations with <tt>mc_prod_test.inp</tt> start MC again from the extended chain starting coordinates and carry out 1M MC steps, using either the unoptimized move set, or the optimized move set (<tt>opt:1</tt>). Two trajectories are written (saving only every 1000th move!), <tt>.data/mc_prod.unopt.dcd</tt> and <tt>.data/mc_prod.opt.dcd</tt>, which can be visualized (together with <tt>./data/rgds.psf</tt> as before).


*Admittedly, while the end results I obtain(ed) are different, it's hard to judge whether the optimized run is better than the unoptimized one.*
