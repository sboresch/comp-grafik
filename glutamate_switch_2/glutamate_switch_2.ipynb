### 1. Propka

The server (link above) is mostly intended for "something else" (preparing the system for a Poisson Boltzmann solver), but it offers the possibility to apply <tt>propka</tt> to the structure and to look at the output. Please do so
for Form III in particular, and see what pKa value is estimated for Glu411 (you should compare to Form II!)

**Optional:** Install Propka yourself!

Propka is nowadays a normal Python package, so installing it is much simpler than it used to be -- there is no need to clone its github repository anymore. Open a terminal window in the JH environment and execute

> pip install --user propka

<font color="blue">**A twist in the new JH environment:**</font> Only your `~/work` directory is persistent; everything else (including wherever `pip install --user` would normally put things) gets wiped when your session ends. To make the installation survive, tell `pip` to install into `~/work` instead, by prefixing the command with the `PYTHONUSERBASE` environment variable, all on one line:

> PYTHONUSERBASE=~/work/pyuser pip install --user propka

(Don't split this into two commands -- `PYTHONUSERBASE=~/work/pyuser` by itself just sets the variable for nothing in particular; it has to be glued to the front of the command it should apply to, as shown.)

`~/work` is also mounted <tt>noexec</tt>, so you cannot run the installed <tt>propka3</tt> command directly by typing its name. Instead, call it through Python explicitly, again with the same `PYTHONUSERBASE=~/work/pyuser` prefix so Python can find what you installed (you'll need to repeat this prefix every session, since only the installed *files* persist, not environment variables):

> PYTHONUSERBASE=~/work/pyuser python3 -c "from propka.run import main; main()" yourfile.pdb

which will produce a file <tt>yourfile.pka</tt> containing the pKa estimates, exactly as the normal <tt>propka3 yourfile.pdb</tt> command would.

<font color="red">**If this does not work:**</font> propka depends on a package (<tt>gemmi</tt>) containing compiled, non-Python code, and I don't yet know whether the `~/work` filesystem permits loading such code at all -- this is one of the things I still need to test in the new environment. If the above fails with an import error, fall back to simply reinstalling propka fresh at the start of every session (just <tt>pip install --user propka</tt>, without the <tt>PYTHONUSERBASE</tt> prefix) -- a bit wasteful, but it only takes a few seconds and sidesteps the issue entirely.

Once this works, you are ready to analyze / estimate the pKa values of amino acids in a protein. I suggest you make a working directory (<tt>mkdir ...</tt>) and change to it (<tt>cd ...</tt>). Then run propka (using whichever of the two invocations above worked for you) on your PDB file; this will produce a <tt>.pka</tt> file containing the info of interest. Again, all this is **optional**, you can also use the webserver!

### 2. Protoss

This server builds optimized hydrogen positions, again, run Form II and Form II through it and download the PDB files it generates -- make sure not to overwrite the initial PDB files!! Take a look at these PDB files using VMD -- how did Protoss handle Glu411 in the two forms? **Note:** The Protoss Webserver can be 'picky'. Please give it a try, but if you it doesn't work for you, don't waste too much time. (Obviously, you can also ask me for help, but recently sometimes failures have been a bit mysterious!)

### 3. How to use the results of these two programs:

From propka you get estimates of the pKa value and then you can use this info in charmm-gui (second screen) to adjust the protonation state of (e.g.) Glu411 (via the pulldown menu)

Using Protoss, you obtain new coordinates with hydrogens. In this case, upload those to charmm-gui (rather than just typing the PDB code). Make sure to click the option "Keep hydrogen atoms". 

### 4. Build a solvated system

Decide on which structure / approach you want to use to continue, then work with charmm-gui "Solution Builder" to build the solvated box. Defaults for most remaining options are fine. Either download just the final PDB file and visualize it, or download the full charmm-gui.tgz (ATTENTION: quite big) and upload it to the JH environment, where you can unpack it. In principle the input files starting with step4 and step5 are ready to run, but need to be adapted for our resources. If you want to give this a try, please get in touch and I'll help you!!
