## Setting up a "real" protein for MD simulation in explicit water

You already started this system as the final task when working with VMD. Here we want to start from (one of the PDB files) and use charmm-gui to set up a simulation system in explicit water. There is likely no the time to run an actual simulation (I would have to transfer the input files to one of our computation nodes), though if you are very quick I can try to help you. This is the final task with less precise documentation, and you'll have to explore a few tools on your own. Obviously, ask for help if you run into issues.

From the VMD exercise you "know" the PDB files, you know what is in there and what constitutes the biological
unit. This is important in the first step/window of charmm-gui. Eventually, you should solvate the system ("Solution builder"), but for trial runs just use "PDB Reader".

Since it is so important, I want you to **pay attention to the protonation state**, specifically to the protonation
state of **Glu 411** in 'Form III'. Please explore two web services to investigate this.

* [propka](http://server.poissonboltzmann.org/pdb2pqr) via a web server (you can also install the command yourself into your JH environment, see below)
* [Protoss](https://proteins.plus/)

### 1. Propka

The server (link above) is mostly intended for "something else" (preparing the system for a Poisson Boltzmann solver), but it offers the possibility to apply <tt>propka</tt> to the structure and to look at the output. Please do so
for Form III in particular, and see what pKa value is estimated for Glu411 (you should compare to Form II!)

**Optional:** Install Propka yourself!

Open a terminal window in the JH environment

Also, point a browser Window to [propka's github repository](https://github.com/jensengroup/propka-3.1). We will mostly follow the intructions there. 

Switch to the terminal window, and excute the command

> git clone https://github.com/jensengroup/propka-3.1.git

You find a new directory <tt>propka-3.1</tt>, change into it, i.e.

> cd propka-3.1

Now, following the instructions on the github site execute

> python3 setup.py install --user

In the JH environment this should just work ("famous last words"). Test it by typing

> propka3

or 

> propka3 --help

The first command does nothing, the second spits out online help. If you get 'command not found', open a new shell window (tab) *or* execute in the shell

> source ~/.profile

and try again.

Once this works, you are ready to analyze / estimate the pKa values of amino acids in a protein. While you can stay in the current directory, I suggest you make a working directory (<tt>mkdir ...</tt>) and change to it (<tt>cd ...</tt>). Then

> propka3 <yourpdb>.pdb
    
which will produce a file <tt>yourpdb.pka</tt>. This contains the info of interest. Again, all this is **optional**, you can also use the webserver!

### 2. Protoss

This server builds optimized hydrogen positions, again, run Form II and Form II through it and download the PDB files it generates -- make sure not to overwrite the initial PDB files!! Take a look at these PDB files using VMD -- how did Protoss handle Glu411 in the two forms? **Note:** The Protoss Webserver can be 'picky'. Please give it a try, but if you it doesn't work for you, don't waste too much time. (Obviously, you can also ask me for help, but recently sometimes failures have been a bit mysterious!)

### 3. How to use the results of these two programs:

From propka you get estimates of the pKa value and then you can use this info in charmm-gui (second screen) to adjust the protonation state of (e.g.) Glu411 (via the pulldown menu)

Using Protoss, you obtain new coordinates with hydrogens. In this case, upload those to charmm-gui (rather than just typing the PDB code). Make sure to click the option "Keep hydrogen atoms". 

### 4. Build a solvated system

Decide on which structure / approach you want to use to continue, then work with charmm-gui "Solution Builder" to build the solvated box. Defaults for most remaining options are fine. Either download just the final PDB file and visualize it, or download the full charmm-gui.tgz (ATTENTION: quite big) and upload it to the JH environment, where you can unpack it. In principle the input files starting with step4 and step5 are ready to run, but need to be adapted for our resources. If you want to give this a try, please get in touch and I'll help you!!
