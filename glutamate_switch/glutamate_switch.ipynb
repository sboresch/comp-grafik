## Making pictures of "real" proteins (working with real world pdb files)

In this directory, you find two pdb files <tt>3v0f.pdb</tt> and <tt>3v0g.pdb</tt> which correspond to "Form II" and "III" of https://doi.org/10.1038/nsmb.2289. A copy of this article (nsmb.2289.pdf) is available on Moodle. Please skim through the paper, you will need it to orient yourself.
The primary difference between the two structures is the orientation of Glu411 (one of the catalytic residues)

Challenges: The PDB files contain multiple copies of the protein; you always need just one! For some residues, there are alternative coordinates. Further, in all of the structures, coordinates for some residues are missing, but these are never the same.

**Tasks:**

- Take one of the structures and make an overview picture (similarly to Fig. 1b of the paper)
- Try to overlay the two structures; you will have to tweak the selections by hand (Aside from different residues missing in the various structures / copies of the protein, the crystal structures contain *alternative* coordinates for some residues. Search for ALTLOC in the VMD documentation)
- Make a "nice" picture where one sees the different "parking positions" of Glu411 (Further, the two structures / forms of the protein are illustrative for pKa shifts (the acidity of Glu411 is likely influenced by its 'parking position'. If there is time, we'll revisit this system using propka ...))
