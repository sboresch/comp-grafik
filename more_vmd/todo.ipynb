## Some additional vmd exercises

In preparation for the final vmd exercise on a voltage dependent
phosphatase ("glutamate_switch"), the following warm up tasks might
help. They all use the protein *triosephosphatase isomerase* (PDB entry: 5TIM). This is one of the "fastest" enzymes, so it has been studied for a long time; 5TIM was an early high resolution structure.

* Try fitting monomer B of 5TIM on monomer A. Watch out, the pdb file contains water and some other "stuff", so the atom selections are not all that trivial; also, it may be tricky to make "all" of monomer B move.
* Interestingly, vmd does not provide a graphical interface to compute the solvent accessible surface area, but there is the <tt>measure sasa</tt> command, see this [entry of the vmd user's guide](http://www.ks.uiuc.edu/Research/vmd/current/ug/node138.html). Working out how this works is a good exercise in vmd scripting. Two postings to the vmd mailing list which I found helpful are [here](http://www.ks.uiuc.edu/Research/vmd/mailing_list/vmd-l/9129.html) and
[here](http://www.ks.uiuc.edu/Research/vmd/mailing_list/vmd-l/8481.html). Start working with a simple protein, such as 1CRN (here the pdb file just contains the protein). Then work your way up to 5TIM -- here it may be helpful to work with an edited copy of the pdb file in which everything but the two protein chains has been removed. Of particular interest is a calculation of the buried surface area upon dimerization, i.e., area(chain A) + area(chain B) - area(dimer).
* Time for some more exercises, inspired by the old [WHATIF](http://swift.cmbi.ru.nl/whatif/) tutorial, all based on the 5TIM pdb file. 
  + If you have not done so, compute the buried surface area of the interface region. 
  + By whatever method, color residues in the interface region of the dimer (there is a <tt>measure</tt> command that might be helpful).
  + Find all waters that are stuck in the interface region
