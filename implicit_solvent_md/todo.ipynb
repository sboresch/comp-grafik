## Implicit solvent MD of a protein

This unit illustrates running a short MD simulation of BPTI (bovine pancreatic trypsin inhibitor) and carrying out basic analyses, i.e., the checks one would carry out to ensure that the simulated protein is "behaving."

For ease of use, all input files are provided. Most input files were generated using the
[CHARMM-GUI implicit solvent modeler](http://www.charmm-gui.org/?doc=input/implicit), and you might want to try going through the steps yourself, choosing FACTS_P22 as the implicit solvent model, and 1bpi as the PDB code. The files are slightly edited / adapted to the purpose of this
course.

The "FACTS" implicit solvent model is described [here](https://onlinelibrary.wiley.com/doi/abs/10.1002/jcc.20832). It is a 'Generalized-Born' type implicit solvent model optimized for speed, providing often a good balance between performance and accuracy. (This setup uses CHARMM22 ("param22"); CHARMM36/36m is the modern default for protein force fields, but for this introductory MD exercise the choice makes no real difference.)



### 1. Work program

(1) Most files in this folder, as mentioned earlier, originate from the 
[CHARMM-GUI implicit solvent modeler](http://www.charmm-gui.org/?doc=input/implicit) utility. I adapted them slightly, important differences will be pointed out. Start by running

    charmm -i step1_pdbreader.inp > step1_pdbreader.out
    
Behind the scenes, the CHARMM-GUI webserver prepared the PDB file for BPTI (code: 1BPI) so that it can be used/parsed by CHARMM. <tt>step1_pdbreader.inp</tt> reads these preprocessed files and sets up the simulation system, i.e., at the end the PSF of the system and starting coordinates in CHARMM format are written. (*Please make sure that you know / still know what the PSF is!*)

Since we will work with an implicit solvent model, all preparatory steps can be carried out in one step (file); in general setting up a simulations system (protein in water, protein in membrane) is a multistep process. The details go beyond what can be taught about CHARMM in this introductory course. The important part is that after running <tt>step1_pdbreader.inp</tt>, we have the final PSF and CHARMM coordinate file from which we can start all further simulations.

(2) CHARMM-GUI also produces an example how to run MD, <tt>step2_implicit.inp</tt>. I have used this as the template for <tt>step2_runmd.inp</tt>, which mostly adapts the number of MD steps, and the MD is divided in an equilibration phase, which is discarded, and a production phase, during which a trajectory is saved. Run

    charmm -i step2_runmd.inp > step2_runmd.out
    
This will take a few minutes. You may want to put the calculation in the "background"

    charmm -i step2_runmd.inp > step2_runmd.out &
    
so that you can use the console; but please watch out that the calculation has finished before continuing.

(3) I further added scripts <tt>step3_anaX.inp</tt>, where X=1,2,3. These carry out some analyses based on the
trajectory written in step (2).  "ana1" and "ana2" represent routine analyses (RMSD, $r_{gyr}$, SASA, secondary structure content) which should be watched during any MD simulation (one could recycle these scripts when working with explicit solvent.) The final script <tt>step3_ana3.inp</tt> carries out an analysis which was looked at in the past: can the tyrosine side chain (of residue 21) "flip" or not, i.e., we compute a time series of the relevant dihedral angle ($\chi_2$).

From a technical point of view, <tt>step3_ana1.inp</tt> and <tt>step3_ana2.inp</tt> both use the [CORRel module](https://www.charmm.org/charmm/documentation/by-version/c42b1/params/doc/correl/). Try to figure out what is going on -- feel free to ask questions.

<tt>step3_ana2.inp</tt> illustrates the second 'main' method of analyzing things which requires a bit of CHARMM "programming". 

One important technical detail: The analysis scripts (which you obviously also have to run with CHARMM) are written in such a way that the regular CHARMM output is not really needed (unless things crash for unexpected reasons). Instead, the scripts each write the pertinent information to separate files, <tt>anaX.dat</tt>. These can be plotted with R (within the JH environment) or downloaded and imported in a spreadsheet etc.

(4) Particularly the analysis scripts may seem confusing, and indeed lots of little details need to be taken care of which require a more detailed understanding of CHARMM than can be conveyed here. However, one positive aspect: These scripts can be trivially adapted to proteins in explict water etc. (The biggest obvious change will be in the respective atom selections, the parts in <tt>SELE..END</tt>. While the syntax is somewhat different, CHARMM's atom selection is as capable as that of VMD.)

**Optional: Running longer simulations:** As written <tt>step2_runmd.inp</tt> carries out a really short number of steps (though this simulation is 3 times longer than the first MD simulation of a protein in 1976!). You can increase the number of steps after the <tt>nstep</tt> keyword in the <tt>dynamics</tt> commands in the file, e.g., set them to 100000 and 200000. In this case, you really want to run in the background!!

*Obviously, you have to rerun the analyses to see what has changed!* (These still run quite fast!)
